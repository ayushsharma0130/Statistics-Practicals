```python
import pandas as pd
from scipy.stats import ttest_ind
import math
from scipy.stats import ttest_ind, f_oneway, chi2_contingency, pearsonr

data = pd.read_csv("/content/evalsdataset.csv")
```

##**Q1. T-Test: Using the teachers' rating data set, does gender affect teaching evaluation rates?**

```python
male = data[data["gender"] == "male"]["score"]
female = data[data["gender"] == "female"]["score"]

# Hypotheses
print("H0: Gender does not affect teaching evaluation rates.")
print("H1: Gender affects teaching evaluation rates.")

# Calculate t-test
t_stat, p_value = ttest_ind(male, female)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
# Decision
alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Gender significantly affects teaching evaluation rates.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Gender does not significantly affect teaching evaluation rates.")
```

##**Q2. ANOVA: Using the teachers' rating data set, does beauty score for instructors differ by age?**

```python
# Hypotheses
print("H0: Beauty score is the same for all age groups.")
print("H1: Beauty score differs for at least one age group.")
```

```python
groups = []

for age in data["age"].unique():
    group = data[data["age"] == age]["bty_avg"]
    groups.append(group)
```

```python
#ANOVA
f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
# Decision
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Beauty score differs significantly by age.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Beauty score does not differ significantly by age.")
```

##**Q3. Chi-square: Using the teachers' rating data set, is there an association between tenure and gender?**

```python
# Hypotheses
print("H0: There is no association between tenure and gender.")
print("H1: There is an association between tenure and gender.")
```

```python
table = pd.crosstab(data["rank"], data["gender"])

print("\nContingency Table:")
print(table)
```

```python
#chi-aquare test
chi2, p_value, dof, expected = chi2_contingency(table)

print("\nChi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

```python
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant association between tenure and gender.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is no significant association between tenure and gender.")
```

##**Q4. Correlation: Using the teachers rating dataset, Is teaching evaluation score correlated with beauty score?**

```python
# Hypotheses
print("H0: There is no correlation between teaching evaluation and beauty score.")
print("H1: There is a correlation between teaching evaluation and beauty score.")
```

```python
#  Pearson correlation
correlation, p_value = pearsonr(
    data["score"],
    data["bty_avg"]
)

print("Correlation coefficient:", correlation)
print("P-value:", p_value)
```

```python
# Decision
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Teaching evaluation score is significantly correlated with beauty score.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Teaching evaluation score is not significantly correlated with beauty score.")
```

**Conclusion
The Pearson correlation coefficient between teaching evaluation score and beauty score is 0.1871, indicating a weak positive correlation. The p-value is 0.0000508, which is less than the significance level of 0.05. Therefore, we reject the null hypothesis (H₀) and conclude that there is a statistically significant relationship between teaching evaluation score and beauty score. However, the relationship is weak, meaning beauty score has only a slight association with teaching evaluation score.**
